```python
import pandas as pd
from pyspark.sql import SparkSession
```

# LIBRARY ->MODULE->FUNTION

```python
df = pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv")
```

```python
spark = SparkSession.builder.getOrCreate()
spark_df = spark.createDataFrame(df)
spark_df.write.format("delta").mode("overwrite").saveAsTable("batch.raw_data.customers")
```

# data load and datawrite 
1. import
- PySpark is an external library
- Python does not know about Spark by default
- Import gives Python access to Spark tools
- Without import → nothing works

2. session creation
- Spark runs on a cluster of machines
- Your Python code runs on driver machine
- Session is the connection between your code and cluster
- Without session → Spark engine is deaf to your code
- Every single operation goes through session

3.DataFrame Creation
- Spark cannot process raw files directly
- File must be loaded into DataFrame first
- Only then you can → filter, join, group, aggregate, transform

- -data farme ?
- A distributed table with rows and columns
- Exactly like a SQL table or Excel sheet
- But data is split across multiple machines in cluster
- Immutable → you never change it, you create new one after every transformation


4.datawrite :
- After loading and transforming → data is in memory only
- Memory is temporary — cluster stops = data lost
- Writing saves data to permanent storage
- Other pipelines, teams, reports can then read and use it

-delta format
- Normal formats like CSV, Parquet are just dumb files
- Delta is a smart storage layer built on top of Parquet
- It adds superpowers that normal files don't have

-mode-?
- When you write data → that path may already have data
- Spark needs to know → what to do with existing data?
- Mode is your instruction to Spark
